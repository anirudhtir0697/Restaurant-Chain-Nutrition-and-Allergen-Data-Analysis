```python
import pandas as pd
import io
from google.colab import files
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats
from scipy.stats import norm
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import seaborn as sns

#Command to import spreadsheet
uploaded = files.upload()
file_name = list(uploaded.keys())[0]
df = pd.read_csv(io.BytesIO(uploaded[file_name]))

def plot_calories(df, restaurant_name, column):
  res_df = df[df['Restaurant'].str.contains(restaurant_name, case=False, na=False)].copy()
  if res_df.empty:
        print(f"Error: '{restaurant_name}' not found. Check your spelling!")
        return
  #Categorizing skew
  mean_val = res_df[column].mean()
  median_val = res_df[column].median()
  skew_diff = mean_val - median_val
  if column == "Calories":
    if abs(skew_diff) <= 50:
        status = "Roughly Symmetric (Honest Menu)"
    elif skew_diff > 50:
        status = "Right Skewed (Calorie Landmines Present)"
    else:
        status = "Left Skewed (Mostly Heavy Items)"
  elif column == "Total Fat (g)" or column == "Carbs (g)" or column == "Protein (g)":
    if abs(skew_diff) <= 5:
        status = "Roughly Symmetric (Honest Menu)"
    elif skew_diff > 5:
        status = "Right Skewed (Calorie Landmines Present)"
    else:
        status = "Left Skewed (Mostly Heavy Items)"
  else:
    if abs(skew_diff) <= 300:
        status = "Roughly Symmetric (Honest Menu)"
    elif skew_diff > 300:
        status = "Right Skewed (Calorie Landmines Present)"
    else:
        status = "Left Skewed (Mostly Heavy Items)"
  if column == "Protein (g)" or column == "Total Fat (g)" or column == "Carbs (g)":
    if abs(skew_diff) <= 6:
        status = "Roughly Symmetric (Honest Menu)"
    elif skew_diff > 6:
        status = "Right Skewed (Calorie Landmines Present)"
    else:
        status = "Left Skewed (Mostly Heavy Items)"
  if column == "Sodium (mg)":
    if abs(skew_diff) <= 300:
        status = "Roughly Symmetric (Honest Menu)"
    elif skew_diff > 300:
        status = "Right Skewed (Calorie Landmines Present)"
    else:
        status = "Left Skewed (Mostly Heavy Items)"


  #Histogram
  print(restaurant_name+": "+status)
  plt.figure(figsize=(10, 6))

  # Determine appropriate binning strategy
  min_val = res_df[column].min()
  max_val = res_df[column].max()
  data_range = max_val - min_val

  if column == "Calories" or column == "Sodium (mg)": # If the data range is smaller than the default binwidth, let seaborn auto-select bins
      sns.histplot(data=res_df, x=column, binwidth=100, kde=False, color='teal', edgecolor='black')
  else: # Otherwise, use the specified binwidth
      sns.histplot(data=res_df, x=column, binwidth=10, kde=False, color='teal', edgecolor='black')

  #Summary stats
  mean_val = res_df[column].mean()
  std_val = res_df[column].std()
  min_val = res_df[column].min()
  q1 = res_df[column].quantile(0.25)
  median_val = res_df[column].quantile(0.50)
  q3 = res_df[column].quantile(0.75)
  max_val = res_df[column].max()
  print(f"Summary Stats for {column}:") # Corrected line
  print(f"Mean: {mean_val:.2f}")
  print(f"Standard Deviation: {std_val:.2f}")
  print(f"Minimum: {min_val}")
  print(f"First Quartile (Q1): {q1}")
  print(f"Median (Q2): {median_val}")
  print(f"Third Quartile (Q3): {q3}")
  print(f"Maximum: {max_val}")
  plt.show() # Added to display plot for plot_calories

#Scatterplot
def create_scatterplot(df, restaurant_name, x_column, y_column):
  plt.figure(figsize=(10, 6))
  res_df = df[df['Restaurant'].str
              .contains(restaurant_name, case=False, na=False)].copy()
  #Line of best fit
  x = res_df[x_column]
  y = res_df[y_column]
  sns.regplot(data=res_df, x=x_column, y=y_column,
                lowess=True,
                color='red',
                line_kws={'color': 'blue', 'linewidth': 3, 'label': 'Auto-Smoothing (LOWESS)'},
                scatter_kws={'edgecolor': 'black', 'alpha': 0.5})

  plt.title(f'{restaurant_name} - {x_column} vs {y_column} (Non-Parametric Fit)')
  plt.xlabel(x_column)
  plt.ylabel(y_column)
  plt.legend()
  plt.show()
def create_violinplot(df, restaurant_list, column):
  res_df = df[df['Restaurant'].isin(restaurant_list)].copy()
  plt.figure(figsize=(12, 7))
  sns.violinplot(data=res_df, x='Restaurant', y=column,
                   hue='Restaurant', palette='muted',
                   inner='quartile', legend=False)
  plt.title(f'{column} Distribution Across Restaurants') # Changed title to be more generic
  plt.xlabel('Restaurant')
  plt.ylabel(column)
  plt.legend(title='Restaurant')
  plt.show()
def bulk_leaderboard(df, top_n=50):
  df_calc = df.copy()
  df_calc['Bulk Score'] = ((df_calc['Protein (g)']) - ((df_calc['Carbs (g)'] - df_calc['Fiber (g)'])*0.5) - (df_calc['Total Fat (g)'] * 0.5) - (df_calc['Sodium (mg)'])/1000 * 1.0 + df_calc['Fiber (g)']*0.5)
  positive_leaderboard = df_calc[df_calc['Bulk Score'] > 0].copy()
  positive_leaderboard = positive_leaderboard.sort_values(by='Bulk Score', ascending=False)
  positive_leaderboard = positive_leaderboard.sort_values(by='Bulk Score', ascending=False)
  display_cols = ['Restaurant', 'Food', 'Protein (g)', 'Bulk Score']
  return positive_leaderboard[display_cols].head(top_n).reset_index(drop=True)
def weightloss_leaderboard(df, top_n=50):
  df_entrees = df[df['Category'] == 'Entree'].copy()
  is_pizza_chain = df_entrees['Restaurant'].isin(["Domino's", "CPK"])
  df_entrees.loc[is_pizza_chain, 'Total Fat (g)'] = df_entrees['Total Fat (g)'] * 3
  df_entrees['Weight Loss Score'] = 20 - df_entrees['Total Fat (g)']
  positive_leaderboard = df_entrees[df_entrees['Weight Loss Score'] > 0].copy()
  positive_leaderboard = positive_leaderboard.sort_values(by='Weight Loss Score', ascending=False)
  display_cols = ['Restaurant', 'Food', 'Total Fat (g)', 'Weight Loss Score']
  return positive_leaderboard[display_cols].head(top_n).reset_index(drop=True)
def hearthealth_leaderboard(df, top_n=20):
  df_entrees = df[df['Category'] == 'Entree'].copy()
  df_entrees['Heart Health Score'] = ((df_entrees['Total Fat (g)'] - df_entrees['Saturated Fat (g)'])*0.1  - (df_entrees['Sodium (mg)']/500 ) + (df_entrees['Fiber (g)'])*0.5 - df_entrees['Saturated Fat (g)']*2)
  positive_leaderboard = df_entrees[df_entrees['Heart Health Score'] > 0].copy()
  positive_leaderboard = positive_leaderboard.sort_values(by='Heart Health Score', ascending=False)
  display_cols = ['Restaurant', 'Food', 'Calories', 'Heart Health Score']
  return positive_leaderboard[display_cols].head(top_n).reset_index(drop=True)
allergens = ['Has_wheat', 'Has_soy', 'Has_milk', 'Has_egg', 'Has_peanut', 'Has_treenut', 'Has_fish', 'Has_shellfish', 'Has_sesame']
allergen_proportion = df[allergens].mean().sort_values(ascending=False)
plt.figure(figsize=(12, 6))
allergen_proportion.plot(kind='bar', color='crimson', edgecolor='black')
plt.title('Hazard Density Map: Indian Cuisine (Top 9 Allergens)', fontsize=14)
plt.ylabel('Proportion of Menu Items (True/Total)', fontsize=12)
plt.xlabel('Allergen Module', fontsize=12)
plt.xticks(rotation=45)
plt.ylim(0, 1) # Sets scale from 0% to 100%
plt.grid(axis='y', alpha=0.3)
plt.show()
def allergen_bargraph(df, restaurant_list):
  top_9 = [
        'Has_milk', 'Has_egg', 'Has_peanut', 'Has_treenut',
        'Has_wheat', 'Has_soy', 'Has_fish', 'Has_shellfish', 'Has_sesame'
  ]
  df_filtered = df[df['Restaurant'].isin(restaurant_list)]
  risk_df = (df_filtered[top_9] > 0).astype(int)
  allergen_proportion = risk_df.mean().sort_values(ascending=False)
  fig, ax = plt.subplots(figsize=(12, 6))
  allergen_proportion.plot(kind='bar', color='crimson', edgecolor='black', ax=ax)

  ax.set_title('Hazard Density Map: Targeted Restaurants (Top 9 Allergens)', fontsize=14)
  ax.set_ylabel('Proportion of Menu Items (True/Total)', fontsize=12)
  ax.set_xlabel('Allergen Module', fontsize=12)
  plt.xticks(rotation=45)
  ax.set_ylim(0, 1)
  ax.grid(axis='y', alpha=0.3)
  plt.tight_layout()

  plt.savefig('hazard_density_map.png')
  plt.show() # Added to display plot for allergen_bargraph
  return allergen_proportion
#Linearize data
def linearize_data(df, restaurant):
  df['ln_Calories'] = np.cbrt(df['Calories'])
  fig, axes = plt.subplots(1, 2, figsize=(12, 5))
  df['Calories'].plot(kind='hist', bins=20, color='coral', edgecolor='black', ax=axes[0])
  axes[0].set_title('Original Distribution (Right-Skewed)', fontsize=14)
  axes[0].set_xlabel('Raw Calories', fontsize=12)
  axes[0].grid(axis='y', alpha=0.3)
  df['ln_Calories'].plot(kind='hist', bins=20, color='teal', edgecolor='black', ax=axes[1])
  axes[1].set_title('Transformed Distribution: ln(Calories)', fontsize=14)
  axes[1].set_xlabel('ln(Calories)', fontsize=12)
  axes[1].grid(axis='y', alpha=0.3)
  plt.tight_layout()
  plt.show()

  df['ln_Fat'] = np.cbrt(df['Total Fat (g)'])
  fig, axes = plt.subplots(1, 2, figsize=(12, 5))
  df['Total Fat (g)'].plot(kind='hist', bins=20, color='coral', edgecolor='black', ax=axes[0])
  axes[0].set_title('Original Distribution (Right-Skewed)', fontsize=14)
  axes[0].set_xlabel('Raw Calories', fontsize=12)
  axes[0].grid(axis='y', alpha=0.3)
  df['ln_Fat'].plot(kind='hist', bins=20, color='teal', edgecolor='black', ax=axes[1])
  axes[1].set_title('Transformed Distribution: ln(Calories)', fontsize=14)
  axes[1].set_xlabel('ln(Calories)', fontsize=12)
  axes[1].grid(axis='y', alpha=0.3)
  plt.tight_layout()
  plt.show()

  df['ln_Protein'] = np.cbrt(df['Protein (g)'])
  fig, axes = plt.subplots(1, 2, figsize=(12, 5))
  df['Protein (g)'].plot(kind='hist', bins=20, color='coral', edgecolor='black', ax=axes[0])
  axes[0].set_title('Original Distribution (Right-Skewed)', fontsize=14)
  axes[0].set_xlabel('Raw Calories', fontsize=12)
  axes[0].grid(axis='y', alpha=0.3)
  df['ln_Fat'].plot(kind='hist', bins=20, color='teal', edgecolor='black', ax=axes[1])
  axes[1].set_title('Transformed Distribution: ln(Calories)', fontsize=14)
  axes[1].set_xlabel('ln(Calories)', fontsize=12)
  axes[1].grid(axis='y', alpha=0.3)
  plt.tight_layout()
  plt.show()

  X = df[['Calories', 'Total Fat (g)', 'Protein (g)', 'Carbs (g)', 'Sodium (mg)']]
  kmeans = KMeans(n_clusters=5, random_state=42)
  df['Cluster'] = kmeans.fit_predict(X)
  items = df[df['Restaurant'] == restaurant]
  print(items[['Restaurant', 'Food', 'Cluster']].head(42))
  cluster_averages = df.groupby('Cluster')[['Calories', 'Protein (g)', 'Total Fat (g)', 'Carbs (g)', 'Sodium (mg)']].mean()
  print(cluster_averages)

  ct = pd.crosstab(df['Restaurant'], df['Cluster'])
  cluster_percentages = ct.div(ct.sum(axis=1), axis=0) * 100
  print(cluster_percentages.round(1))



def allergen_heatmap(df):
  top_9 = [
        'Has_milk', 'Has_egg', 'Has_peanut', 'Has_treenut',
        'Has_wheat', 'Has_soy', 'Has_fish', 'Has_shellfish', 'Has_sesame'
    ]
  risk_df = df[top_9].isin([1, 2]).astype(int)
  risk_df['Restaurant'] = df['Restaurant']
  heatmap_data = risk_df.groupby('Restaurant')[top_9].mean() * 100
  fig, ax = plt.subplots(figsize=(14, 8))
  sns.heatmap(
          heatmap_data,
          annot=True,
          fmt=".1f",
          cmap='YlOrRd',
          linewidths=0.5,
          ax=ax
      )

  ax.set_title('Allergen Hazard Map (% of Menu Items Containing Allergen)', fontsize=16, pad=15)
  ax.set_xlabel('Allergen Type', fontsize=12)
  ax.set_ylabel('Restaurant Chain', fontsize=12)

  plt.tight_layout()
  plt.savefig('restaurant_allergen_heatmap.png')
  plt.show()

  return heatmap_data









#Display output
restaurants = ["McDonald's", "Burger King", "Wendy's", "Chick-fil-A", "Popeyes", "Taco Bell", "Chipotle", "Subway", "Panera", "IHOP", "Olive Garden", "Chili's", "Noodles & Company", "Red Lobster", "Dunkin", "Starbucks", "Domino's", "Panda Express", "CPK"]
for restaurant in restaurants:
  print(restaurant)
  plot_calories(df, restaurant, "Calories")
  plot_calories(df, restaurant, "Total Fat (g)")
  plot_calories(df, restaurant, "Carbs (g)")
  plot_calories(df, restaurant, "Protein (g)")
  plot_calories(df, restaurant, "Sodium (mg)")
  create_scatterplot(df, restaurant, "Calories", "Total Fat (g)")
  create_scatterplot(df, restaurant, "Calories", "Carbs (g)")
  create_scatterplot(df, restaurant, "Calories", "Protein (g)")
  create_scatterplot(df, restaurant, "Calories", "Sodium (mg)")
  allergen_bargraph(df, [restaurant]) # Fix: Pass restaurant as a list
  linearize_data(df, restaurant)
create_violinplot(df, ["McDonald's", "Burger King", "Wendy's"], "Calories")
create_violinplot(df, ["McDonald's", "Burger King", "Wendy's"], "Total Fat (g)")
create_violinplot(df, ["McDonald's", "Burger King", "Wendy's"], "Carbs (g)")
create_violinplot(df, ["McDonald's", "Burger King", "Wendy's"], "Protein (g)")
create_violinplot(df, ["McDonald's", "Burger King", "Wendy's"], "Sodium (mg)")
create_violinplot(df, ["Chili's", "Taco Bell", "Chipotle"], "Calories")
create_violinplot(df, ["Chili's", "Taco Bell", "Chipotle"], "Total Fat (g)")
create_violinplot(df, ["Chili's", "Taco Bell", "Chipotle"], "Carbs (g)")
create_violinplot(df, ["Chili's", "Taco Bell", "Chipotle"], "Protein (g)")
create_violinplot(df, ["Chili's", "Taco Bell", "Chipotle"], "Sodium (mg)")
create_violinplot(df, ["Popeyes", "Chick-fil-A"], "Calories")
create_violinplot(df, ["Popeyes", "Chick-fil-A"], "Total Fat (g)")
create_violinplot(df, ["Popeyes", "Chick-fil-A"], "Carbs (g)")
create_violinplot(df, ["Popeyes", "Chick-fil-A"], "Protein (g)")
create_violinplot(df, ["Popeyes", "Chick-fil-A"], "Sodium (mg)")
create_violinplot(df, ["Olive Garden", "Red Lobster", "IHOP"], "Calories")
create_violinplot(df, ["Olive Garden", "Red Lobster", "IHOP"], "Total Fat (g)")
create_violinplot(df, ["Olive Garden", "Red Lobster", "IHOP"], "Carbs (g)")
create_violinplot(df, ["Olive Garden", "Red Lobster", "IHOP"], "Protein (g)")
create_violinplot(df, ["Olive Garden", "Red Lobster", "IHOP"], "Sodium (mg)")
create_violinplot(df, ["Dunkin", "Starbucks"], "Calories")
create_violinplot(df, ["Dunkin", "Starbucks"], "Total Fat (g)")
create_violinplot(df, ["Dunkin", "Starbucks"], "Carbs (g)")
create_violinplot(df, ["Dunkin", "Starbucks"], "Protein (g)")
create_violinplot(df, ["Dunkin", "Starbucks"], "Sodium (mg)")
create_violinplot(df, ["Domino's", "CPK"], "Calories")
create_violinplot(df, ["Domino's", "CPK"], "Total Fat (g)")
create_violinplot(df, ["Domino's", "CPK"], "Carbs (g)")
create_violinplot(df, ["Domino's", "CPK"], "Protein (g)")
create_violinplot(df, ["Domino's", "CPK"], "Sodium (mg)")
create_violinplot(df, ["Panda Express", "Noodles & Company"], "Calories")
create_violinplot(df, ["Panda Express", "Noodles & Company"], "Total Fat (g)")
create_violinplot(df, ["Panda Express", "Noodles & Company"], "Carbs (g)")
create_violinplot(df, ["Panda Express", "Noodles & Company"], "Protein (g)")
create_violinplot(df, ["Panda Express", "Noodles & Company"], "Sodium (mg)")
create_violinplot(df, ["Arby's", "Subway", "Panera"], "Calories")
create_violinplot(df, ["Arby's", "Subway", "Panera"], "Total Fat (g)")
create_violinplot(df, ["Arby's", "Subway", "Panera"], "Carbs (g)")
create_violinplot(df, ["Arby's", "Subway", "Panera"], "Protein (g)")
create_violinplot(df, ["Arby's", "Subway", "Panera"], "Sodium (mg)")


leaderboard = bulk_leaderboard(df)
print(leaderboard)
leaderboard1 = weightloss_leaderboard(df)
print(leaderboard1)
leaderboard2 = hearthealth_leaderboard(df)
print(leaderboard2)
allergen_heatmap(df)
```
